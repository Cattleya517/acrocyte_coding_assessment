## Background

Grayscale images are stored as 2D matrices. Pixel value = intensity; higher value = brighter pixel. The images here are **12-bit** (new scope, max 4095) and **16-bit** (old scope, max 65535).

Immunofluorescence (IF) staining lets us see protein markers. Each slide is stained with four colors — **blue** (nucleus), **red** (protein X), **green** (protein Y), **yellow** (protein Z). You don't need to memorize which color is which protein — there are simply four channels.

The same slide was scanned with two microscopes: **old** (16-bit) and **new** (12-bit). So we have 4 channels × 2 scans = 8 images, plus a **mask** file per scan. The mask labels every cell with a unique integer; pixels with value 0 are background.

**Autofluorescence:** even without staining, cells fluoresce a bit. So "does this cell express protein X?" is not just "is its red intensity nonzero?". Raw intensities have no units — they only mean something *relative to other intensities measured under the same conditions*.

Data layout (after the download cell):
```
data/
├── old microscope/   old blue|green|red|yellow|mask .tif
├── new microscope/   new blue|green|red|yellow|mask .tif
└── cells csv/        old cells.csv, new cells.csv
```

## Demo — how `vmin`/`vmax` affect `imshow` display

`imshow` maps array values to greyscale: pixels at `vmax` show as white, pixels at `vmin` (or below) show as black. Same data, different `vmax` → different look. This matters for 12-bit vs 16-bit images later.

````python
im = np.array([
    [0, 1, 2, 1, 0],
    [1, 1, 2, 1, 1],
    [0, 1, 2, 1, 0],
])

fig, ax = plt.subplots(1, 2, figsize=(10, 3))
ax[0].set_title('vmax = 2')
ax[0].imshow(im, cmap='grey', vmin=0, vmax=2)
ax[1].set_title('vmax = 4')
ax[1].imshow(im, cmap='grey', vmin=0, vmax=4)
plt.show()
````
