## Unlock the problem pool

Downloads a tiny encrypted file (a few KB) and decrypts it with the password your interviewer gives you. The password box does not echo; it is not saved in this notebook.

```python
URL = "https://github.com/Cattleya517/acrocyte_coding_assessment/releases/download/data-v1/leetcode_pool_encrypted.zip"
ENC = "leetcode_pool_encrypted.zip"

import os, getpass, urllib.request, pyzipper
if not os.path.exists(ENC):
    urllib.request.urlretrieve(URL, ENC)

pw = getpass.getpass("Data password (ask interviewer): ").encode()
with pyzipper.AESZipFile(ENC) as zf:
    zf.setpassword(pw)
    zf.extractall()  # -> ./leetcode_pool.txt
print("Pool unlocked.")
```

## Draw your warmup problem

Run this cell once. (You can re-run to draw again — but your interviewer is watching, so the first draw is the one that counts.)

```python
import random
from IPython.display import display, HTML

urls = [l.strip() for l in open("leetcode_pool.txt")
        if l.strip() and not l.startswith("#")]
problem = random.choice(urls)

print("Your warmup problem:")
print("   ", problem)
display(HTML(
    f'<a href="{problem}" target="_blank" rel="noopener" '
    'style="font-size:16px">▶ Open this problem on leetcode.com</a>'
))
```
